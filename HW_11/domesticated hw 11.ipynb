### Задание 1
Печатные газеты использовали свой формат дат для каждого выпуска. Для каждой газеты из списка напишите формат указанной даты для перевода в объект `datetime`:

1. The Moscow Times - Wednesday, October 2, 2002
2. The Guardian - Friday, 11.10.13
3. Daily News - Thursday, 18 August 1977

### Пример работы программы
Программа должна выводить на экран объекты типа `datetime`, соответствующие датам в условии задачи.

```python
from datetime import datetime

# Заданные строки с датами
date_moscow_times_str = "Wednesday, October 2, 2002"
date_guardian_str = "Friday, 11.10.13"
date_daily_news_str = "Thursday, 18 August 1977"

# Преобразование строк в datetime объекты с помощью strptime()
dt_moscow_times = datetime.strptime(date_moscow_times_str, "%A, %B %d, %Y")
dt_guardian = datetime.strptime(date_guardian_str, "%A, %d.%m.%y")
dt_daily_news = datetime.strptime(date_daily_news_str, "%A, %d %B %Y")

# Вывод результатов
print(f"The Moscow Times: {dt_moscow_times}")
print(f"The Guardian: {dt_guardian}")
print(f"Daily News: {dt_daily_news}")
```

# Задание 2 (Необязательное)
Напишите функцию `data_range()`, которая возвращает список дат за период от `start_date` до `end_date`. Даты должны вводиться в формате `YYYY-MM-DD`. В случае неверного формата или при `start_date > end_date` должен возвращаться пустой список.


```python
from datetime import datetime, timedelta

# Функция для генерации списка дат
def data_range(start_date_str, end_date_str):
    try:
        # Преобразование строковых дат в datetime объекты
        start_date = datetime.strptime(start_date_str, '%Y-%m-%d')
        end_date = datetime.strptime(end_date_str, '%Y-%m-%d')

        # Если start_date больше end_date, возвращаем пустой список
        if start_date > end_date:
            return []

        # Генерация списка дат
        dates = []
        current_date = start_date
        while current_date <= end_date:
            dates.append(current_date.strftime('%Y-%m-%d'))
            current_date += timedelta(days=1)
        return dates
    except ValueError:
        # В случае ошибки формата даты возвращаем пустой список
        return []

# Ввод пользователем данных
start_date = input('Введите start_date (YYYY-MM-DD): ')
end_date = input('Введите end_date (YYYY-MM-DD): ')

# Получаем список дат и выводим результат
dates = data_range(start_date, end_date)
if dates:
    print('Сгенерированные даты:')
    for date in dates:
        print(date)
else:
    print('Неверный формат даты или start_date > end_date.')
```
