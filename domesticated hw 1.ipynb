#### Задание 1: Определение центральных символов в строке

```python
# Задача: извлечь центральные символы из введенного слова
print('=== Анализатор центральных символов ===')

# Получаем строку от пользователя
user_input = input('Введите произвольное слово: ')

# Выводим исходные данные
print(f'Анализируемое слово: {user_input}')

# Определяем длину строки
str_len = len(user_input)

# Находим центральные символы в зависимости от четности длины
if str_len % 2 == 0:
 # Для четной длины берем два центральных символа
 center_chars = user_input[str_len//2 - 1: str_len//2 + 1]
else:
 # Для нечетной длины берем один центральный символ
 center_chars = user_input[str_len//2]

# Выводим результат
print(f'Центральные символы: {center_chars}')
```

#### Задание 2: Формирование пар из двух списков

```python
# Списки участников
male_participants = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
female_participants = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

print('=== Система подбора пар ===
')

# Проверяем соответствие размеров списков
if len(male_participants) != len(female_participants):
 print('Ошибка: количество участников не совпадает!')
 print('Некоторые участники останутся без пары.')
else:
 # Сортируем оба списка для последовательного сопоставления
 male_participants_sorted = sorted(male_participants)
 female_participants_sorted = sorted(female_participants)
 
 print('Сформированные пары участников:')
 
 # Используем enumerate для нумерации пар
 for i, (boy, girl) in enumerate(zip(male_participants_sorted, 
 female_participants_sorted), 1):
 print(f'{i}. {boy} + {girl}')
```
