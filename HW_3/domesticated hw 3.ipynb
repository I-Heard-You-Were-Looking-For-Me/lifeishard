### Задание 1
Переведите содержимое файла `purchase_log.txt` в словарь `purchases` вида:
`{'1840e0b9d4': 'Продукты', …}`

Пример работы программы при выводе первых двух элементов словаря `purchases`:
```
1840e0b9d4 'Продукты'
4e4f90fcfb 'Электроника'
```


````python
import json

# Инициализация словаря для хранения данных
purchases = {}

# Чтение данных из файла purchase_log.txt
with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    next(f)  # Пропускаем первую строку заголовка
    for line in f:  # Проходим по строкам файла
        purchase = json.loads(line)  # Декодируем строку в JSON
        purchases[purchase['user_id']] = purchase['category']  # Заполняем словарь

# Запрос количества элементов для вывода
a = input('Сколько выводим элементов? ')
print('User ID    Category')

for i, (user_id, category) in enumerate(purchases.items()):
    if i < int(a):  # Ограничение по числу выводимых элементов
        print(f'{user_id} {category}')
    else:
        break  # Прерываем цикл после вывода нужного количества элементов
````

### Задание 2
Для каждого `user_id` в файле `visit_log.csv` определите третий столбец с категорией покупки.
Если покупка была, сам файл `visit_log.csv` изменять не надо.

Запишите в файл `funnel.csv` визиты из файла `visit_log.csv`, в которых были покупки с указанием категории.

Учтите условия для данных:
• содержимое `purchase_log.txt` помещается в оперативную память;
• содержимое `visit_log.csv` — нет, используйте только построчную обработку этого файла.

Пример работы программы при выводе первых трёх строк файла `funnel.csv`:
```
user_id,source,category
1840e0b9d4,other,Продукты
4e4f90fcfb,context,Электроника
```


````python
import json
import csv

# Чтение purchase_log.txt и формирование словаря покупок
purchases = {}
with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    next(f)  # Пропускаем заголовок
    for line in f:
        purchase = json.loads(line.strip())  # Преобразуем строку в JSON
        purchases[purchase['user_id']] = purchase['category']  # Наполняем словарь

# Работа с файлом visit_log.csv и запись в funnel.csv
with open('visit_log.csv', 'r', encoding='utf-8') as in_file:
    with open('funnel.csv', 'w', encoding='utf-8') as out_file:
        writer = csv.writer(out_file)
        reader = csv.DictReader(in_file)
        writer.writerow(['user_id', 'source', 'category'])  # Запись заголовков
        for row in reader:
            user_id = row['user_id'].strip()  # Извлечение user_id
            if user_id in purchases:
                category = purchases[user_id]
                writer.writerow([user_id, row['source'], category])  # Запись строки в funnel.csv
                print(f'Дописано: {user_id}, {row['source']} => {category}')  # Сообщение для отладки

# Вывод первых трёх строк файла funnel.csv
with open('funnel.csv', 'r', encoding='utf-8') as f:
    for i in range(3):  # Вывод первых трёх строк
        print(f.readline().strip())
````
